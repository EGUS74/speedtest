```python
import speedtest

def start_speed_test():
    wifi = speedtest.Speedtest()
    wifi.get_best_server()

    download_speed = wifi.download() / 1_000_000  # Convert to Mbps

    upload_speed = wifi.upload() / 1_000_000  # Convert to Mbps
    
    # return f"Download Speed: {download_speed:.2f} Mbps", f"Upload Speed: {upload_speed:.2f} Mbps"
    return download_speed, upload_speed
```

```python
start_speed_test()
```
